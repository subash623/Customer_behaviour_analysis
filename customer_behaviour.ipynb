```python
# fill the null value with median
# 1. Using median over mean coz outlier may present which effect the data.
# 2. Group by category so that it will calculate the value in the same category and fill up with median
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# All the column is in Capital letter and have a unnecessary space
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ",'_')
```

```python
df.columns
```

```python
df=df.rename(columns ={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
df.columns
```

```python
# Create a column age_group
labels = {'Young Adult','Adult','Middle_aged', 'Senior'}
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
```

```python
# create column purchase_frequency_days
frequency_mapping ={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,	
    'Annually':365,
    'Bi-Weekly':14,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
